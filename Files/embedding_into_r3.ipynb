```python
from mpl_toolkits.mplot3d import Axes3D
import numpy as np
import matplotlib.pyplot as plt
```

```python
def plot_3d_vectors(vectors):
    fig = plt.figure()
    ax = fig.add_subplot(111, projection='3d')

    vectors_np = np.array(vectors)

    # Plot as scatter points (dots)
    ax.scatter(
        vectors_np[:, 0], vectors_np[:, 1], vectors_np[:, 2],
        color='b', s=50  # s controls the size of the dots
    )

    # Scale based on min and max values from all vectors
    min_val = vectors_np.min()
    max_val = vectors_np.max()
    
    # Add small padding for better visualization
    padding = (max_val - min_val) * 0.1
    
    ax.set_xlim([min_val - padding, max_val + padding])
    ax.set_ylim([min_val - padding, max_val + padding])
    ax.set_zlim([min_val - padding, max_val + padding])
    ax.set_xlabel('X')
    ax.set_ylabel('Y')
    ax.set_zlabel('Z')

    # Set initial viewing angle
    ax.view_init(elev=20, azim=45)  # elevation and azimuth angles

    plt.show()
```

```python
embedding_300d = np.random.rand(300)
# print(embedding_300d)
# split the embedding into groups of 3
vectors_3d = embedding_300d.reshape(-1, 3)

plot_3d_vectors(vectors_3d)
# Softmax function to convert logits to probabilities

def softmax(logits):
    exp_logits = np.exp(logits - np.max(logits))  # Subtract max for numerical stability
    return exp_logits / exp_logits.sum(axis=-1, keepdims=True)

# Example usage

converted = softmax(embedding_300d)
splitted = converted.reshape(-1, 3)
plot_3d_vectors(splitted)

```
